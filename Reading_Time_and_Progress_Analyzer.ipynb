```python
def calculate_reading_speed(pages, minutes):
  if minutes == 0:
    return 0
  return pages / minutes

def classify_speed(speed):
  if speed < 1:
    return "Developing"
  elif speed <= 2:
    return "Steady"
  else:
    return "Fast"

sessions = int(input("How many reading sessions? "))

total_pages = 0
total_minutes = 0

for i in range(1, sessions + 1):
  pages = int(input(f"\nSession {i} pages read: "))
  minutes = float(input("Minutes spent: "))

  speed = calculate_reading_speed(pages, minutes)

  print(f"Reading speed: {speed:.2f} pages/minute")
  print(f"Classification: {classify_speed(speed)}")

  total_pages += pages
  total_minutes += minutes

overall_speed = calculate_reading_speed(total_pages, total_minutes)

print(f"\nTotal pages read: {total_pages}")
print(f"Total reading time: {total_minutes:g} minutes")
print(f"Overall reading speed: {overall_speed:.2f} pages/minute")
```
