```python
def update_savings(current_total, deposit):
  return current_total + deposit

def check_goal(total, target):
  if total >= target:
    return "Goal reached!"
  else:
    return "Keep saving!"

target = float(input("Enter savings target: ₱"))
total = 0

while True:
  deposit = float(input("Enter deposit (0 to finish): ₱"))

  if deposit == 0:
    break

  total = update_savings(total, deposit)
  print(f"Updated savings: ₱{total:.2f}")

print(f"\nTotal savings: ₱{total:.2f}")

if total < target:
  print(f"Remaining amount needed: ₱{target - total:.2f}")

print(check_goal(total , target))

```
