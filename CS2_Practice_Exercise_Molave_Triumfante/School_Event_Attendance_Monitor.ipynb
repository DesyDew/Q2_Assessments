```python
def calculate_attendance_rate(attendees, expected):
  if expected == 0:
    return 0
  return (attendees / expected) * 100

def classify_attendance(rate):
  if rate < 60:
    return "Low"
  elif rate < 90:
    return "Moderate"
  else: return "High"

sessions = int(input("How many sessions? "))

total_attendees = 0
total_expected = 0
high_sessions = 0

for i in range(1, sessions + 1):
  expected = int(input(f"\nSession {i} expected attendance: "))
  attendees = int(input("Actual attendees: "))

  rate = calculate_attendance_rate(attendees, expected)
  classification = classify_attendance(rate)

  print(f"Attendance rate: {rate:.2f}%")
  print(f"Classification: {classification}")

  if classification == "High":
    high_sessions += 1

  total_attendees += attendees
  total_expected += expected

overall_rate = calculate_attendance_rate(total_attendees, total_expected)

print(f"\nTotal attendees: {total_attendees}")
print(f"Total expected attendance: {total_expected}")
print(f"Overall attendance rate: {overall_rate:.2f}%")
print(f"High-attendance sessions: {high_sessions}")
```
