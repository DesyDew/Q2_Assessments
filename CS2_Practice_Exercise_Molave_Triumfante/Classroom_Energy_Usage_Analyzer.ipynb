```python
def calculate_energy(power_watts, hours_used):
  return (power_watts * hours_used) / 1000

def classify_usage(energy):
  if energy < 0.5:
    return "Low"
  elif energy <= 2.0:
    return "Moderate"
  else:
    return "High"

num_devices = int(input("How many device? "))

total_energy = 0
highest_energy = -1
highest_device = ""

for i in range(1, num_devices + 1):
  name = input(f"\nDevice {i} name: ")
  power = float(input("Power rating (watts): "))
  hours = float(input("Hours used: "))

  energy = calculate_energy(power, hours)
  classification = classify_usage(energy)

  print(f"Energy consumption: {energy:.2f} kWh")
  print(f"Classification: {classification}")

  total_energy += energy

  if energy > highest_energy:
    highest_energy = energy
    highest_device = name

print(f"\nTotal energy consumption: {total_energy:.2f} kWh")

if num_devices > 0:
  print(f"Highest consumption device: {highest_device}")
```
