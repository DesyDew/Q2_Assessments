```python
def calculate_sale(quantity, price_per_container):
  return quantity * price_per_container

def calculate_discount(total):
  if total >= 500:
    return total * 0.05
  else:
    return 0

price = float(input("Price per container: ₱"))

total_sales = 0
total_discounts = 0

while True:
  quantity = int(input("\nEnter quantity sold (0 to finish): "))

  if quantity == 0:
    break

  sale = calculate_sale(quantity, price)
  discount = calculate_discount(sale)
  amount_due = sale - discount

  print(f"Sale: ₱ {sale:.2f}")
  print(f"Discount: ₱ {discount:.2f}")
  print(f"Amount due: ₱ {amount_due:.2f}")

  total_sales += sale
  total_discounts += discount

print(f"\nTotal sales before discounts: ₱ {total_sales:.2f}")
print(f"Total discounts: ₱ {total_discounts:.2f}")
print(f"Net sales: ₱ {total_sales - total_discounts:.2f}")
```
